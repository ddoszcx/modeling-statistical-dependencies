```python
import pandas as pd


df = pd.read_csv('data.csv', sep = '\t')
df

```

```python
import numpy as np



X = df.drop(columns = ['y'])
y = df['y']

mnk_beta = np.linalg.inv(X.T @ X)  @ X.T @ y
mnk_beta

```

```python

a = 0.1
ridge_beta = np.linalg.inv(X.T @ X + a * np.eye((X.T @ X).shape[0])) @ X.T @ y




ridge_beta
```

```python
from sklearn.linear_model import Lasso

from sklearn.model_selection import train_test_split

# suppress=True убирает 1e-05, precision=4 задает 4 знака после запятой
np.set_printoptions(suppress=True, precision=4)




lasso = Lasso(alpha= 0.1)
lasso.fit(df.drop(columns = ['y']), df['y'])
pd.DataFrame(lasso.coef_)
```

В исходной модели МНК коэффициенты сильно искажены из-за мультиколлинеарности: признаки x1, x2 и x3 тесно связаны между собой (корреляция между x1 и x3 достигает 0.98). В результате дисперсия оценок раздулась, а коэффициент при x2 получил неадекватное и контринтуитивное значение −18.38, из-за чего классический МНК нельзя корректно интерпретировать.

Ridge-регуляризация (L2) решила проблему раздутых весов за счет их стягивания к нулю. Оценка при x2 сократилась с −18.38 до приемлемых −2.69, сохранив при этом все признаки в модели и сделав ее устойчивой к шуму.

Lasso-регуляризация (L1) пошла дальше и выполнила автоматический отбор признаков. Она полностью занулила избыточные и дублирующие факторы (x2, x3, x5). В итоговой модели остались только два ключевых фактора: x4 (коэффициент около 2.28) и x1 (коэффициент около 0.34), которые вносят основной положительный вклад в целевую переменную y.
